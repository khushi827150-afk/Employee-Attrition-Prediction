Multivariate Analysis

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Convert Attrition into numerical values
df['Attrition_Numeric'] = df['Attrition'].map({'Yes': 1, 'No': 0})

# Select numerical columns
numerical_df = df.select_dtypes(include='number')

# Remove constant columns
numerical_df = numerical_df.loc[:, numerical_df.nunique() > 1]

# Calculate correlation matrix
corr_matrix = numerical_df.corr()

# Plot heatmap
plt.figure(figsize=(18, 14))

sns.heatmap(
    corr_matrix,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    linewidths=0.5,
    center=0,
    vmin=-1,
    vmax=1
)

plt.title('Correlation Heatmap of Employee Attrition Dataset', fontsize=16)
plt.tight_layout()
plt.show()
```

Data Preprocessing

```python
pd.options.display.max_columns = 500
```

```python
df
```

```python
#Binary = Attrition, Gender, Over 18, OverTime

df["Attrition"] = df["Attrition"].apply(lambda x: 1 if x == 'Yes' else 0)
df["Gender"] = df["Gender"].apply(lambda x: 1 if x == 'Yes' else 0)
df["Over18"] = df["Over18"].apply(lambda x: 1 if x == 'Yes' else 0)
df["OverTime"] = df["OverTime"].apply(lambda x: 1 if x == 'Yes' else 0)
```

```python
df
```

```python
# One-Hot Encoding: Business Travel, Department, EducationField, JobRole, Marital Status

df = df.join(pd.get_dummies(df['BusinessTravel'])).drop('BusinessTravel', axis=1)

df = df.join(pd.get_dummies(df['Department'], prefix='Department')).drop('Department', axis=1)

df = df.join(pd.get_dummies(df['EducationField'], prefix='Education')).drop('EducationField', axis=1)

df = df.join(pd.get_dummies(df['JobRole'], prefix='Role')).drop('JobRole', axis=1)

df = df.join(pd.get_dummies(df['MaritalStatus'], prefix='Status')).drop('MaritalStatus', axis=1)
```

```python
df
```

```python
df = df.map(lambda x: 1 if x is True else 0 if x is False else x)
```

```python
df = df.drop('EmployeeNumber', axis=1)
```

```python
import matplotlib.pyplot as plt
df.hist(figsize=(20, 15))
plt.tight_layout()
plt.show()
```

```python
df = df.drop(['EmployeeCount', 'Over18', 'StandardHours'], axis=1)
```

```python
df
```

Model Training

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
X, y = df.drop('Attrition', axis=1), df['Attrition']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
model = RandomForestClassifier(n_jobs=-1)
model.fit(X_train, y_train)
```

Model Evaluation

```python
model.score(X_test, y_test)
```

Faeture Importance

```python
sorted_importances = dict(sorted(zip(model.feature_names_in_, model.feature_importances_), key=lambda x: x[1], reverse=True))
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(20, 8))
plt.bar(sorted_importances.keys(), sorted_importances.values())
plt.xticks(rotation=45, ha='right')
plt.show()
```

```python
import pandas as pd
import plotly.express as px
from IPython.display import display, Markdown

# Use the existing dataset
# Assumes your cleaned dataset is stored in a DataFrame called df

total_employees = len(df)

employees_left = (
    df['Attrition']
    .astype(str)
    .str.strip()
    .str.lower()
    .eq('yes')
    .sum()
)

employees_stayed = total_employees - employees_left

attrition_rate = (
    employees_left / total_employees * 100
    if total_employees > 0 else 0
)

display(Markdown("## 📊 Employee Overview"))

display(Markdown(
    f"""
| Metric | Value |
|---|---:|
| Total Employees | {total_employees:,} |
| Employees Who Left | {employees_left:,} |
| Employees Who Stayed | {employees_stayed:,} |
| Attrition Rate | {attrition_rate:.2f}% |
"""
))
```
